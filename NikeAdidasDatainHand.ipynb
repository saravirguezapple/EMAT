### How has online search interest in Nike and Adidas changed in the United States from 2021 to 2025? ###

## Data Source and Collection

For this project, I used Google Trends to collect data about Nike and Adidas. I searched for both brands and selected the United States as the location, with a time range from January 2021 to December 2025.

I kept the search settings on All categories and Web Search. After that, I downloaded the information from the Interest over time section as a CSV file.

The data shows how much people searched for each brand over time, using values from 0 to 100. These numbers don't represent the exact number of searches, but they help me compare the search interest for both brands.

```python
import pandas as pd

df = pd.read_csv("multiTimeline.csv", skiprows=2)

print("Dataset size:", df.shape)

df.head()
```

## Understanding My Data

My dataset has 262 rows and 3 columns. Each row represents one week, and the columns show the date and the search interest for Nike and Adidas.

The three columns are: 
Week: The date when each week starts.
Nike: The search interest for Nike during that week.
Adidas: The search interest for Adidas during that week.

The dates help me see how the searches changed over time, while the numbers help me compare both brands. I also noticed that the first date is December 27, 2020, because Google Trends uses the starting date of each week.

```python
df = df.rename(columns={
    "Nike: (United States)": "Nike",
    "Adidas: (United States)": "Adidas"
})

df["Week"] = pd.to_datetime(df["Week"])

df["Nike"] = pd.to_numeric(df["Nike"])
df["Adidas"] = pd.to_numeric(df["Adidas"])

print("Data types:")
print(df.dtypes)

df.head()
```

## Checking My Data

Before doing any analysis, I wanted to check if there was anything missing or wrong with my data.

I checked for missing values and repeated dates to make sure the information was organized correctly. I also checked how many weeks were included in the dataset.

This is important because I want to make sure I'm working with the right information before comparing the two brands.

```python
print("Missing values:")
print(df.isnull().sum())

print("\nRepeated dates:")
print(df["Week"].duplicated().sum())

print("\nFirst and last dates:")
print(df["Week"].min())
print(df["Week"].max())

print("\nTotal weeks:")
print(len(df))
```

## Comparing the Numbers

Now that I have my data organized, I wanted to get a general idea of what the numbers look like for each brand.

I'm going to look at the average search interest, along with the highest and lowest values. This will help me understand if one brand has been searched more often than the other and give me a starting point for the next part of my project.

```python
print("Summary of Nike and Adidas:")
display(df[["Nike", "Adidas"]].describe().round(2))

print("\nAverage search interest:")

print("Nike:", round(df["Nike"].mean(), 2))
print("Adidas:", round(df["Adidas"].mean(), 2))
```

## What I Found and What's Next

After looking at my data, I noticed that Nike had a higher average search interest than Adidas during the time period I selected. Nike had an average of around 61.76, while Adidas was around 19.03.

I also checked the dataset and found that there were no missing values, which is good because I can continue working with the information I already have.

Something I want to keep in mind is that Google Trends only shows search interest, so this doesn't necessarily mean that one brand sells more than the other.
